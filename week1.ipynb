```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.array([1, 8])
y = np.array([3, 10])

plt.plot(x, y)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]
plt.figure(figsize=(8,5))
plt.plot(x,y,color='darkgreen',linestyle=':',marker='s',linewidth=3)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("line chart")
plt.legend(['my data series'])
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[5,2,3,4,6]
plt.figure(figsize=(8,5))
plt.bar(categories,values,color='#23054A')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar chart')
plt.show()
```

```python

# Sample data
data = [12, 15, 13, 17, 16, 16, 19, 12, 14, 16, 14, 17, 18, 18, 15, 14]

# Create histogram
plt.hist(data, bins=16, color='#850109')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Simple Histogram Example')
plt.show()
```

```python
# Sample data
x = [1, 2, 3, 4, 5]
y = [5, 7, 4, 6, 8]

# Create scatter plot
plt.scatter(x, y)
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Simple Scatter Plot Example')
plt.show()
```

```python
# Sample data
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]

# Create pie chart
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Simple Pie Chart Example')
plt.axis('equal')  # Ensures the pie is drawn as a circle
plt.show()
```

```python

# Sample data
data = [7, 8, 5, 6, 7, 9, 12, 5, 7, 8, 6]

# Create box plot
plt.boxplot(data)
plt.title('Simple Box Plot Example')
plt.ylabel('Values')
plt.show()
```

```python

# Sample data: 10x10 matrix
data = np.random.rand(10, 10)

# Create heatmap
plt.imshow(data, cmap='viridis', interpolation='nearest')
plt.colorbar()  # Show color scale
plt.title('Simple Heatmap Example')
plt.show()
```
